```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
%matplotlib inline
```

```python
def sigmoid_01(x, k, midpoint):
    sig = lambda t: 1 / (1 + np.exp(-t))
    lo = sig(-k * midpoint)
    hi = sig(k * (1 - midpoint))
    return (sig(k * (x - midpoint)) - lo) / (hi - lo)

x = np.linspace(0, 1, 500)

def plot_sigmoid(k=8.0, midpoint=0.5):
    fig, ax = plt.subplots(figsize=(7, 5))
    ax.plot(x, sigmoid_01(x, k, midpoint), color='steelblue', lw=2, label='sigmoid')
    ax.plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--', label='linear (y = x)')
    ax.plot(midpoint, 0.5, 'o', color='tomato', zorder=5, label=f'midpoint ({midpoint:.2f}, 0.5)')
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xlabel('input x')
    ax.set_ylabel('f(x)')
    ax.set_title(f'Sigmoid [0,1] → [0,1]   k={k},  midpoint={midpoint:.2f}')
    ax.legend(loc='upper left')
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

interact(
    plot_sigmoid,
    k=FloatSlider(value=8, min=0.5, max=20, step=0.5, description='Steepness (k)'),
    midpoint=FloatSlider(value=0.5, min=0.05, max=0.95, step=0.01, description='Midpoint'),
)
```

```python
def powered(x, power_of):

    return x ** power_of

x = np.linspace(0, 1, 500)

def plot_sigmoid(power=1):
    fig, ax = plt.subplots(figsize=(7, 5))
    ax.plot(x, powered(x, power), color='steelblue', lw=2, label='sigmoid')
    ax.plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--', label='linear (y = x)')
    # ax.plot(midpoint, 0.5, 'o', color='tomato', zorder=5, label=f'midpoint ({midpoint:.2f}, 0.5)')
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.set_xlabel('input x')
    ax.set_ylabel('f(x)')
    ax.set_title(f'X to the Power of midpoint={power:.2f}')
    ax.legend(loc='upper left')
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

interact(
    plot_sigmoid,
    power=FloatSlider(value=8, min=0.25, max=4, step=0.1, description='Power'),
)
```
